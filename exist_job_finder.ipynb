```python
import pandas as pd
import os
import pathlib
```

### Open csv dir and load all csv

```python
def load_data_csv(path):
    if not os.path.exists(path):
        raise ValueError("Error: path does not exist")
    
    try:
        for root, _, files in os.walk(path):
            for f in files:
                if f.endswith(".csv"):
                    yield os.path.join(root, f)
    except Exception as e:
        raise ValueError(f"Error loading files due to: {e}")
```

```python
try:
    all_dfs = []
    for csv_file in load_data_csv("./csv_dir"):
        df = pd.read_csv(csv_file)
        all_dfs.append(df)
    ads = pd.concat(all_dfs, ignore_index=True)
    print(ads.head())
except Exception as e:
    print(f"error occur load dataset due to: {e}")
```

```python
ads.drop_duplicates(subset=['title', 'company', 'location', 'description','url'], inplace=True)
ads.dropna(inplace=True)
ads
 
```

```python
def check_duplicate(job, df_existing):
    # Create a unique identifier (tuple) for the job based on key fields
    job_identifier = (
        job["title"],
        job["company"],
        job["location"],
        job["description"],
        job["url"]
    )
    # Convert the DataFrame into a set of unique job identifiers (tuples)
    df_existing_identifiers = set(df_existing[['title', 'company', 'location', 'description', 'url']].apply(tuple, axis=1))

    return job_identifier in df_existing_identifiers
```

```python
def filter_unique_jobs(job_data_list, df_existing):
    unique_jobs = []
    for job in job_data_list:
        if   check_duplicate(job, df_existing):
            unique_jobs.append(job)
    return unique_jobs
```
